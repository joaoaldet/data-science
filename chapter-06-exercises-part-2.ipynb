```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Exercício 1)

```python
iris = sns.load_dataset('iris')
# print(iris.columns)

setosa = iris[iris['species'] == 'setosa']
corr = setosa[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr()

sns.heatmap(
    corr,
    annot = True,
    fmt = '.2f'
)

plt.show()
```

Exercício 2)

```python
titanic = sns.load_dataset('titanic')
# print(titanic.columns)

sns.histplot(
    data = titanic,
    x = 'age',
    hue = 'sex',
    kde = True
)

plt.title('Distribuição das idades dos passageiros por sexo')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.legend(title = 'Sexo', labels = ['Masculino', 'Feminino'])

plt.show()
```

Exercício 3)

```python
sns.boxplot(
    data = titanic,
    x = 'pclass',
    y = 'age',
    hue = 'sex'
)

plt.title('Distribuição das idades por classe e sexo')
plt.xlabel('Classe')
plt.ylabel('Idade')
plt.legend(title = 'Sexo')

plt.show()
```

Exercício 4)

```python
mpg = sns.load_dataset('mpg')
# print(mpg.columns)

sns.regplot(
    data = mpg,
    x = 'horsepower',
    y = 'mpg',
    scatter_kws = {'color': 'grey', 'alpha': 0.6},
    line_kws = {'color': 'red'}
)

plt.title('Relação entre potência e consumo de combustível')
plt.xlabel('Potência (horsepower)')
plt.ylabel('Consumo (mpg)')
plt.show()
```

Exercício 5)

```python
ds = pd.read_csv('paises.csv', delimiter = ';')

corr = ds[['GDP ($ per capita)',
           'Literacy (%)',
           'Infant mortality (per 1000 births)',
           'Phones (per 1000)']].corr()

sns.heatmap(
    corr,
    annot = True,
    fmt = '.2f'
)

plt.show()
```

Exercício 6)

```python
# print(ds.columns)

dados = ds[
    ds['Region'].str.contains('LATIN AMER. & CARIB') |
    ds['Region'].str.contains('WESTERN EUROPE')
]

sns.boxplot(
    data = dados,
    x = 'Region',
    y = 'GDP ($ per capita)'
)

plt.title('GDP per capita: América Latina x Europa Ocidental')
plt.xlabel('Região')
plt.ylabel('GDP ($ per capita)')
plt.show()
```

Exercício 7)

```python
sns.regplot(
    data = ds,
    x = 'Literacy (%)',
    y = 'Infant mortality (per 1000 births)',
    line_kws = {'color': 'red'},
     scatter_kws = {'s': 20}
)

plt.title('Relação entre alfabetização e mortalidade infantil')
plt.xlabel('Taxa de alfabetização (%)')
plt.ylabel('Mortalidade infantil (por 1000 nascimentos)')

plt.show()
```

Exercício 8)

```python
ds_space = pd.read_csv('space.csv', delimiter = ';')

space = ds_space[ds_space[' Cost'] > 0]

dados = space[
    space['Status Rocket'].str.contains('StatusActive') |
    space['Status Rocket'].str.contains('StatusRetired')
]

sns.boxplot(
    data = dados,
    x = 'Status Rocket',
    y = ' Cost'
)

plt.xlabel('Status do foguete')
plt.ylabel('Custo (milhões de US$)')

plt.show()
```
