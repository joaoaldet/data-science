```python
import numpy as np

dataset = np.loadtxt('space.csv', delimiter = ';', dtype = 'str')
print(dataset[0:])
```

Exercício 1)

```python
status = dataset[1:, 7]

sucesso = status[status == 'Success']

porcen = len(sucesso) / len(status) * 100
print('A porcentagem de sucesso das missões foi de', porcen, '%')
```

Exercício 2)

```python
gastos = dataset[1:, 6].astype(float)
custospos = gastos[gastos > 0]
media = np.mean(custospos)
print('Média de gastos das missões:', media, 'milhões de dólares.')
```

Exercício 3)

```python
company = dataset[1:, 2]

mask = np.char.find(company, 'USA') >= 0
print('Foram feitas pelos Estados Unidos', mask.sum(), 'missões.')
```

Exercício 4)

```python
company = dataset[1:, 1]
num = dataset[1:, 0]
gastos = dataset[1:, 6].astype(float)

mask_spacex = company == 'SpaceX'
custos = gastos[mask_spacex]
num_spacex = num[mask_spacex]
i_maior = np.argmax(custos)
print('A missão mais cara da SpaceX foi a de número:', num_spacex[i_maior])
```

Exercício 5)

```python
empresas, quantidades = np.unique(dataset[1:, 1], return_counts = 'True')

for empresa, qtd in zip(empresas, quantidades):
    print(empresa, '-', qtd)
```

Exercício 6)

```python
status = dataset[1:, 5]

retired = status[status == 'StatusRetired']

porcen = len(retired) / len(status) * 100
print('A porcentagem de missões com "StatusRetired" é de', porcen, '%')
```

```python
company = dataset[1:, 2]

mask = np.char.find(company, 'Russia') >= 0
print('Foram feitas na Rússia', mask.sum(), 'missões.')
```

```python
company = dataset[1:, 1]
gastos = dataset[1:, 6].astype(float)

i_maior = np.argmax(gastos)
print('A missão mais cara foi feita pela', company[i_maior], 'com custo de', gastos[i_maior], 'milhões de dólares.')
```
