```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Exercício 1)

```python
ds = np.loadtxt('paises.csv', delimiter=';', dtype='str')

regioes = ds[:, 1]
mask = np.char.find(regioes, 'NORTHERN AMERICA') >= 0
america = ds[mask]

x = america[:, 0]
y = america[:, 15].astype(float)
y2 = america[:, 16].astype(float)

plt.plot(x, y, 'r', linewidth=3)
plt.plot(x, y2, 'b', linewidth=3)

plt.show()
```

Exercício 2)

```python
ds = np.loadtxt('space.csv', delimiter=';', dtype='str')

mask_usa = np.char.find(ds[:, 2], 'USA') >= 0
mask_china = np.char.find(ds[:, 2], 'China') >= 0

empresas_eua = np.unique(ds[mask_usa, 1])
empresas_china = np.unique(ds[mask_china, 1])

x = ['USA', 'China']
y = [len(empresas_eua), len(empresas_china)]

plt.bar(x, y)

plt.xlabel('País')
plt.ylabel('Número de empresas')

plt.show()
```

Exercício 3)

```python
ds = np.loadtxt('space.csv', delimiter=';', dtype='str')

mask_roscosmos = np.char.find(ds[:, 1], 'Roscosmos') >= 0
roscosmos = ds[mask_roscosmos]

sucesso = np.sum(roscosmos[:, 7] == 'Success')
falha = np.sum(roscosmos[:, 7] == 'Failure')

x = ['Sucesso', 'Falha']
y = [sucesso, falha]

plt.pie(y, labels=x, autopct='%1.1f%%')

plt.xlabel('Resultado das missões')
plt.show()
```

Exercício 4)

```python
ds = pd.read_csv('space.csv', delimiter=';')

failure = ds[ds.iloc[:, 7] == 'Failure']

empresas = failure.iloc[:, 1].value_counts().head(5)

plt.bar(empresas.index, empresas.values)

plt.xlabel('Empresas')
plt.ylabel('Número de missões com Failure')

plt.show()
```

Exercício 5)

```python
ds = pd.read_csv('paises.csv', delimiter=';')

america_latina = ds[ds['Region'].str.contains('LATIN AMER.')]

plt.scatter(
    america_latina['GDP ($ per capita)'],
    america_latina['Literacy (%)'],
    s=america_latina['Population'] / 1000000)

plt.xlabel('GDP ($ per capita)')
plt.ylabel('Literacy (%)')

plt.show()
```

Exercício 6)

```python
ds = pd.read_csv('space.csv', delimiter=';')

foguetesAtivos = ds[ds['Status Rocket'] == 'StatusActive']

qtdAtivos = len(foguetesAtivos)
qtdAposentados = len(ds) - qtdAtivos

plt.pie(
    x=[qtdAtivos, qtdAposentados],
    labels=['% Foguetes ativos', '% Foguetes aposentados'],
    autopct='%1.1f%%'
)

plt.show()
```

Exercício 7)

```python
ds = pd.read_csv('paises.csv', delimiter=';')

europa = ds[ds['Region'].str.contains('WESTERN EUROPE')]

x = range(len(europa))

y = europa['GDP ($ per capita)']
y2 = europa['Phones (per 1000)']

plt.plot(x, y, 'ro--')
plt.plot(x, y2, 'bs-')

plt.xlabel('Países')
plt.ylabel('Valor')

plt.show()
```

Exercício 8)

```python
ds = pd.read_csv('space.csv', delimiter=';')

sucesso = ds[ds['Status Mission'] == 'Success']
falha = ds[ds['Status Mission'] == 'Failure']

empresasSucesso = sucesso['Company Name'].value_counts().head(5)
empresasFalha = falha['Company Name'].value_counts().head(5)

plt.subplot(1, 2, 1)
plt.bar(empresasSucesso.index, empresasSucesso.values)
plt.xticks(rotation = 45)
plt.title('Missões de sucesso')

plt.subplot(1, 2, 2)
plt.bar(empresasFalha.index, empresasFalha.values)
plt.xticks(rotation = 45)
plt.title('Missões de falha')

plt.show()
```
