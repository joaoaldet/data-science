```python
import numpy as np

arr1 = np.ones(8)
arr2 = np.random.randint(0, 10, 8)
arr3 = arr1 + arr2
print('Array:', arr3)

soma = arr3.sum(axis=0)
print('Soma:', soma)

if(soma >= 40):
  arr4 = arr3.reshape(4, 2)
else:
  arr4 = arr3.reshape(2, 4)

print('Matriz:', arr4)
```

Exercício 3)

```python
mtz = np.zeros([2,2])
blin = np.random.randint(0,2)
bcol = np.random.randint(0,2)
mtz[blin, bcol] = 1

jogadas = []
bomba = False

while len(jogadas) < 3 and not bomba:
  lin = int(input('Digite a linha (1 ou 2):')) - 1
  col = int(input('Digite a coluna (1 ou 2):')) - 1

  if lin < 0 or lin > 1 or col < 0 or col > 1:
    print('Posição inválida! Digite 1 ou 2.')
    continue

  if mtz[lin, col] == 1:
    print('Game Over! :( Try Again!')
    bomba = True
  else:
    if [lin, col] not in jogadas:
      jogadas.append([lin, col])
    print('Keep playing.')

if not bomba and len(jogadas) == 3:
  print('Congratulations! You beat the game! :)')
```

```python
np.random.seed(10)
mtz = np.random.randint(0, 51, [4, 4])
linha = mtz.mean(axis=1)
coluna = mtz.mean(axis=0)
print('Média das linhas:', linha)
print('Média das colunas:', coluna)
print('Maior valor da média da linha:', linha.max())
print('Maior valor da média da coluna:', coluna.max())
elem, cont = np.unique(mtz, return_counts= True)
print('Elementos presentes:', elem)
print('Quantidade de aparições:', cont)
num2x = elem[cont == 2]
print('Números que aparecem duas vezes:', num2x)
```
