Exercício 6)

```python
dec = float(input('Digite um número decimal:'))
print('Raíz quadrada:',math.sqrt(dec))
print('Função teto:',math.ceil(dec))
print('Função chão:',math.floor(dec))
print('Parte inteira:',math.trunc(dec))
```

Exercício 7)

```python
p = str(input('Digite uma palavra:'))
n = len(p)
pm = p.upper()
for c in range(0,n):
  print(pm[c])
la = pm.count('A')
le = pm.count('E')
li = pm.count('I')
lo = pm.count('O')
lu = pm.count('U')
vogais = la + le + li + lo + lu
print('O número de vogais da palavra é:', vogais)
if('A' in pm):
  print('A letra A está presente na palavra.')
```

```python
n1 = float(input('Digite um número:'))
n2 = float(input('Digite outro número:'))

print('Adição:', n1 + n2)
print('Subtração:', n1 - n2)
print('Multiplicação:', n1 * n2)
print('Divisão:', n1 / n2)
print('Resto da divisão:', math.fmod(n1, n2))
print('Potência:', math.pow(n1, n2))
```
