```python
import numpy as np
import pandas as pd
```

Exercício 1)

```python
ds = pd.read_csv('paises.csv', sep=';')
cond = ds['Region'].str.contains('OCEANIA')

print('Países da Oceania:')
print(ds[cond]['Country'])

qtd = len(ds[cond])
print(qtd, 'países são da Oceania.')
```

Exercício 2)

```python
print('País com a maior população:')
print(ds.nlargest(1, 'Population')['Country'])

print('Região do país:')
print(ds.nlargest(1, 'Population')['Region'])
```

Exercício 3)

```python
group_region = ds.groupby('Region')
media = group_region['Literacy (%)'].mean()

print('Média de alfabetização de cada região em %:')
print(media.round(2))
```

Exercício 5)

```python
dr = ds['Deathrate']

def deathRate(dr):
  if dr < 9:
    return 'Balanced'
  else:
    return 'Urgent'

ds['Humanitarian Help'] = ds['Deathrate'].apply(deathRate)

print(ds)
```

Exercício 6)

```python
group_region = ds.groupby('Region')
print(group_region.describe()['Population'])
```

Exercício 7)

```python
mortality1 = ds['Infant mortality (per 1000 births)']

def reducedMortality(mortality1):
  return mortality1 * 0,85

mortality2 = ds['Infant mortality (per 1000 births)'].apply(reducedMortality)
mortality2.name = ''
pd.concat([mortality1, mortality2], axis=1)
```
