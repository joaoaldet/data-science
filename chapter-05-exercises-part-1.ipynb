```python
import numpy as np
import pandas as pd
```

Exercício 1)

```python
seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})
```

Exercício 2)

```python
print('Porcentagem total que as linguagens do Ano 1 representam juntas:', np.sum(seriesAno1), '%')
print('Porcentagem total que as linguagens do Ano 2 representam juntas:', np.sum(seriesAno2), '%')
```

Exercício 3)

```python
crescimento = seriesAno2 - seriesAno1
print('Crescimento/declínio no mercado:')
print(crescimento)
```

Exercício 4)

```python
cond = crescimento > 0
print('Linguagens que tiveram crescimento:')
print(crescimento[cond])
```

Exercício 5)

```python
cresc_dois = seriesAno2 + crescimento * 2
print('Linguagem mais popular em dois anos:')
print(cresc_dois.nlargest(1))
```

Exercício 6)

```python
np.random.seed(10)

df = pd.DataFrame(
    index = ['A', 'B', 'C', 'D', 'E'],
    columns=['W', 'X', 'Y', 'Z'],
    data=np.random.randint(1, 50, [5, 4]))

cond = df['X'] < 30
media = df['X'][cond].mean()
print('Média dos elementos da coluna X que são menores que 30:', media)
```

Exercício 7)

```python
media_d = df.loc['D'].mean()
print('Média da linha D:', media_d)

soma_e = df.iloc[4, :].sum()
print('Soma dos elementos da linha E:', soma_e)
```

Exercício 8)

```python
mtz = df.loc[['A', 'C', 'E'], ['X', 'Y']]

print('Soma dos elementos das linhas:')
print(mtz.sum(1))
print('Soma dos elementos das colunas:')
print(mtz.sum())
```
