```python
loja1 = {'Galaxy S24FE', 'Galaxy A57', 'iPhone 17 Pro', 'Moto G86'}
loja2 = {'Galaxy S25', 'iPhone 17 Pro', 'Moto G52'}

print('A loja 1 tem', len(loja1), 'modelos disponíveis.')
print('A loja 2 tem', len(loja2), 'modelos disponíveis.')

comuns = list(set(loja1) & set(loja2))

print('Ambas as lojas possuem os seguintes modelos:', comuns)
```

Exercício 4)

```python
nomes = []
pesos = []

for i in range(3):
  nome = input('Informe seu nome:')
  peso = input('Informe seu peso:')
  nomes.append(nome)
  pesos.append(peso)

maior = max(pesos)
menor = min(pesos)

i_maior = pesos.index(maior)
i_menor = pesos.index(menor)
print('A pessoa de maior peso é', nomes[i_maior])
print('A pessoa de menor peso é', nomes[i_menor])
```

Exercício 5)

```python
nomes = []
idades = []
sexos = []
menor20 = 0
soma = 0

n = int(input('Digite o número de pessoas:'))
for i in range(n):
  nome = input('Informe seu nome:')
  idade = int(input('Informe sua idade:'))
  sexo = input('Informe seu sexo:').lower()
  nomes.append(nome)
  pesos.append(peso)
  sexos.append(idade)
  soma += idade
  if idade < 20 and sexo == 'f':
    menor20 += 1

media = soma / n
print('Média das idades:', media)
print((menor20), 'mulher(es) tem menos de 20 anos.')
```

Exercício 8)

```python
lista = []

for i in range(3):
  nome = input('Informe o produto:')
  preco = float(input('Informe o preço:'))
  estoque = int(input('Informe o estoque:'))
  dic = {'nome': nome,
        'preco': preco,
        'estoque': estoque}
  lista.append(dic)

for produto in lista:
  print('Nome do produto:', produto['nome'])
  print('Valor total:', produto['preco'] * produto['estoque'])
```
